## Exercise 1: Function Plotting and Analytical Determination of Critical Points

Determine _analytically_ the coordinates and nature of the critical points of the following functions:

- $$f_1(x,y) = x^3 + y^3 -3x -12y$$
- $$f_2(x,y) = x^3 -12xy + 8y^3$$
- $$f_3(x,y) = {(x-1)}^2 + 10 {(x^2-y)}^2$$

**_Vocabulary:_** An "analytical" determination means that the calculation is done "by hand", without using numerical tools or formal computation.

```python
# Import numpy and pyplot libraries
import numpy as np
import math 
import matplotlib.pyplot as plt
```

```python
def f1(x, y):
    return x**3 + y**3 - 3*x - 12*y

def f2(x, y):
    return x**3 - 12*x*y + 8*y**3

def f3(x, y):
    return (x - 1)**2 + 10 * (x**2 - y)**2
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math

# define f1
def f1(x, y):
    return x**3 + y**3 - 3*x - 12*y

#  critical points of f1 (computed analytically) 
critical_points = [
    (1, 2),
    (1, -2),
    (-1, 2),
    (-1, -2)
]

# plot domain
xmin, xmax, nx = -4, 4, 200
ymin, ymax, ny = -4, 4, 200

x1d = np.linspace(xmin, xmax, nx)
y1d = np.linspace(ymin, ymax, ny)
x2d, y2d = np.meshgrid(x1d, y1d)

# contour plot
plt.figure(figsize=(6,6))
nIso = 35
plt.contour(x2d, y2d, f1(x2d, y2d), nIso, cmap='viridis')

# plot critical points 
for (xc, yc) in critical_points:
    plt.plot(xc, yc, 'ro', markersize=8)
    plt.text(xc+0.1, yc+0.1, f"({xc},{yc})", color='red')

plt.title("Contour Lines of f₁(x,y) with Critical Points")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.axis("square")
plt.show()
```

**3.3** The instruction block below provides the template for the function _df1_, which calculates and returns the gradient of $f_1$. Complete this instruction block so that _df1_ does what is expected. Then use this function to verify that the gradient of $f_1$ is indeed zero at the critical points.

```python
def df1(x, y):
    # partial derivative with respect to x
    df_dx = 3 * x**2 - 3
    
    # partial derivative with respect to y
    df_dy = 3 * y**2 - 12
    
    return [df_dx, df_dy]

# critical points of f1
critical_points = [
    (1, 2),
    (1, -2),
    (-1, 2),
    (-1, -2)
]

# check that gradient = 0 at each point 
for (xc, yc) in critical_points:
    print(f"Gradient at ({xc}, {yc}) =", df1(xc, yc))
```

**3.4** The instruction block below provides the template for the function _d2f1_, which calculates and returns the Hessian matrix of $f_1$. Complete this instruction block so that _d2f1_ does what is expected. Then use this function to verify the nature of the critical points of $f_1$.

```python
# Instruction block to be completed
def d2f1(x, y):
    # Second-order partial derivatives
    d2f_dxdx = 6 * x
    d2f_dxdy = 0
    d2f_dydx = 0
    d2f_dydy = 6 * y
    
    return [[d2f_dxdx, d2f_dxdy],
            [d2f_dydx, d2f_dydy]]
```

```python
# Template for calculating eigenvalues of a matrix, to be adapted:
matrice = [[1,2],[3,4]]
valProp = np.linalg.eigvals(matrice)
print('Eigenvalues of the matrix {} :'.format(matrice))
print('  vp1 = {}'.format(valProp[0]))
print('  vp2 = {}'.format(valProp[1]))
```

```python
# To be completed to determine the nature of the critical points of f1
# Determine the nature of the critical points of f1

critical_points = [(1,2),(1,-2),(-1,2),(-1,-2)]

for (xc,yc) in critical_points:
    H = d2f1(xc,yc)
    eigvals = np.linalg.eigvals(H)
    
    print(f"\nPoint ({xc},{yc})")
    print("Hessian =", H)
    print("Eigenvalues =", eigvals)
    
    if eigvals[0] > 0 and eigvals[1] > 0:
        print(" --> Local MINIMUM")
    elif eigvals[0] < 0 and eigvals[1] < 0:
        print(" --> Local MAXIMUM")
    else:
        print(" --> SADDLE point")

# Calculate the Hessian matrix
matrice1 = d2f1 (-1,-2)
# Calculate the eigenvalues of the Hessian matrix, then test their sign
valProp = np.linalg.eigvals(matrice1)
print('Eigenvalues of the matrix {} :'.format(matrice1))
print('  vp1 = {}'.format(valProp[0]))
print('  vp2 = {}'.format(valProp[1]))
```
